```python
marks=[
    int(input("Marks in English:")),
    int(input("Marks in Science:")),
    int(input("Marks in Maths:"))
]
if any(m<35 for m in marks):
    print("You failed the exam")
else:
    total=sum(marks)
    per=total/len(marks)

    if per>=80:
        grade="A"
    elif per>=60:
        grade="B"
    else:
        grade="C"
    print(f"You passed with {per:.2f}% in Grade {grade}")    
```
