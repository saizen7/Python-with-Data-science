```python
import matplotlib.pyplot as plt
import seaborn as sns
data=[12,15,14,10,8,7,13]
sns.boxplot(data=data)
plt.title("Box plot")
plt.ylabel("Values")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
data=[12,15,14,10,8,7,13]
sns.violinplot(data=data)
plt.title("violin plot")
plt.ylabel("Values")
```
