```python
# arithmetic

arr = np.array([1,2,3])

print(arr + 5)   # [6 7 8]
print(arr * 2)   # [2 4 6]
print(arr - 3)   # [-2 -1  0]

print("Original array:", arr)  # [1 2 3]
```

```python
#write a program to calculate add sub mul div and sqr of an array in numpy also filter the data 
#of each variable taking its value show data greater than mean value

arr = np.array([1,2,3,4,5])
result = arr + 2
print(result)
result = arr - 3
print(result)
result = arr * 5
print(result)
result = arr / 2
print(result)
result = arr ** 1
print(result)

print("Mean result ")

mean_val = np.mean(arr)
print(mean_val)

print("Filter result")
filtered = arr[arr > mean_val]
print(filtered)
```

```python
a= np.array([1,2,3,4,5])

print(f"The mean is {np.mean(a)}")
print(f"The median is {np.median(a)}")
print(f"The standard is {np.std(a)}")
print(f"The  variance is {np.var(a)}")
print(f"The min is {np.min(a)}")
print(f"The max is {np.max(a)}")
```

```python
import numpy as np

arr = np.array([2,4,6,7,8,6,4])
brr = np.array([34,56,57,67,89,45,23])

print(np.percentile(arr,10))
print(np.quantile(arr,0.5))
print(np.corrcoef(arr,brr))  # correlation
```
