```python
import csv

print("=== STEP 1: READ THE CSV FILE ===")

with open("Customers-100.csv", "r") as file:
    reader = csv.DictReader(file)
    customers = list(reader)

print(f"Total Records Loaded: {len(customers)}")

print("Sample Record:")

# check if data exists before accessing index 0
if len(customers) > 0:
    print(customers[0])
else:
    print("No data found in CSV file")
```

```python
print("\n==== Step 2: CLEAN DATA ====")

#convert index to integer
for c in customers:
    c["Index"] = int(c["Index"])

print("Converted 'Index' to integer.")
print(type(customers[0]["Index"]))
```

```python
print("\n==== Step 3: CLEAN DATA ====")

#get first 5 names
names=[c["First Name"] for c in customers]
print("FIrst 5 Names",names[:5])
```

```python
#coutries
countries = [c["Country"]for c in customers]
print ("First 5 countries:",countries[:5])
```

```python
print("\n==== STEP 4: FILTER (Customers from Chile) ====")

chile_customers = list(filter(lambda c: c["Country"] == "Chile",customers))
print(f"Total Chile Customers: {len(chile_customers)}")
print("Sample:",chile_customers)
```

```python
print("\n==== STEP 5: FILTER (Recent Customers after 2020) ====")

recent= list(filter(lambda c: c["Subscription Date"] > "2020-12-31", customers))
print(f"Recent Customers Count: {len(recent)}")
```
