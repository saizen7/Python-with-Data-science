```python
students = [] # declaring a variable list to store values

n = int(input("Enter number of students (more than 5): "))

subjects = ["English","Math","Nepali","Science"]

for i in range(n):
    #print("\nEnter details for Student", i + 1)
    name = input("Enter name: ")
    marks = []
    # Input marks using nested loop
    for sub in subjects:
        mark = int(input(f"Enter marks in {sub}: "))
        marks.append(mark)

    total = sum(marks)
    percentage = total / len(subjects)

    # Fail condition
    if any(m < 32 for m in marks):
        status = "Fail"
        grade  = "No Grade"
    else:
        status = "Pass"

        if percentage >= 80:
            grade = "A+"
        elif percentage >= 70:
            grade = "A"
        elif percentage >= 60:
            grade = "B"
        elif percentage >= 50:
            grade = "C"
        elif percentage >= 40:
            grade = "D"
        else:
            grade = "F"
    students.append({
        "name": name,
        "English": marks[0],
        "Math": marks[1],
        "Nepali": marks[2],
        "Science": marks[3],
        "Total": total,
        "Percentage": percentage,
        "Status": status,
        "Grade": grade
    })


print(students)

    # ==== RESULT SHEET ====

print("\n\n ========== STUDENT RESULT SHEET ==========\n")

for s in students:
    print("Name         :", s["name"])
    print("English      :", s["English"])
    print("Math         :", s["Math"])
    print("Nepali       :", s["Nepali"])
    print("Science      :", s["Science"])
    print("Total        :", s["Total"])
    print("Percentage   :", round(s["Percentage"],2))
    print("Status       :", s["Status"])

    # Grade only if PASS
    if s["Status"] == "Pass":
        print("Grade      :",s["Grade"])
    else:
        print("Grade      : No Grade (Fail)")

    print("-" * 40)
```

```python
students = {
    "Ram": {"Maths":80, "Science":70, "English":90},
    "Shyam": {"Maths":90, "Science":85, "English":88},
    "Hari": {"Maths":60, "Science":65, "English":70}
}
for name,subjects in students.items():
    total=sum(subjects.values())
    avg=total/len(subjects)

    if avg>=80:
        grade="A"
    elif avg>=60:
        grade="B"
    else:
        grade="C"
    print(f"{name}: Average: {avg:.2f},Grade:{grade}")
```

```python
cart={
    "apple":{"price":100,"qty":2},
    "banana":{"price":80,"qty":5},
    "milk":{"price":80,"qty":1}
}

total = 0

for item, details in cart.items():
    cost = details["price"] * details["qty"]
    total += cost
    print(item, ":",cost)
print("Total Bill:",total)
```
