```python
# A funtion to clean
# This function should remove all None values and print the clean data 
def clean(data):
    result=[]
    for item in data:
        if type(item)==int or type(item)==float:
            result.append(item)
    return tuple(result)
data=(10,None,30,"No",40,50)
print(clean(data))
    
```

```python
def login_system(user_role):
    print("Checking login credentials....")
    #nested funciton: only accessible after login

    def access_dashboard():
        print("Welocme to dashboard")
    #permission check

    if user_role== "admin":
        print("Login Successful")
        access_dashboard() #inner function runs only of allowed
    else:
        print("Access Denied")
login_system("admin")
login_system("Guest")
```

```python
#Square list numbers map() applies a funciton to every item in a list.


nums=[1,2,3,4]

result =list(map(lambda x: x*x, nums))
print(result)
```

```python
# Convert string to uppercase
words = ["apple", "mango", "orange"]

uppercase = list(map(lambda x: x.upper(), words))

print(uppercase)

# Convert string to lowercase
words = ["APPLE", "MANGO", "ORANGE"]

lowercase = list(map(lambda x: x.lower(), words))

print(lowercase)
```

```python
# Convert Celsius to Fahrenheit
celsius = [0, 20, 30]

fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```

```python
def factorial(n):
    if n==0  or n==1:
        return 1
    return n*factorial(n-1)
print(factorial(5))
```

```python
def sum_list(lst):
    if len(lst)==0:
       return 0
    return lst[0]+sum_list(lst[1:])
print(sum_list([1,2,3,4]))
```

```python
def fibonacci(n):
    if n<= 1:
        return n
    return fibonacci(n-1)+ fibonacci(n-2)
print(fibonacci(6))
```
