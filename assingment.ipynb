```python
import numpy as np

a = np.array([1, 2, 3, 4, 5], dtype=np.int32)
b = np.array([1.5, 2.3, 3.7, 4.1], dtype=np.float64)
c = np.array([True, False, True, False], dtype=bool)
d = np.array(['2024-05-01', '2024-05-18', '2025-12-25'], dtype='datetime64')

print(a,a.dtype)
print(b,b.dtype)
print(c,c.dtype)
print(d,d.dtype)
```
